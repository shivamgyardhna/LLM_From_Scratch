```python
import torch
import torch.nn as nn
import pandas as pd
```

```python
torch.manual_seed(123)
```

```python
batch_example = torch.randn(2,5)
layer = nn.Sequential(nn.Linear(5,6), nn.ReLU())
out = layer(batch_example)

print(out)
```

```python
mean = out.mean(dim=-1, keepdim=True) #dim=-1 last dim which is col her and keepdim to keep the dim
var = out.var(dim=-1, keepdim=True)
```

```python
print(mean)
print(var)
```

```python
out_Norm=(out-mean)/torch.sqrt(var)
```

```python
print(out_Norm.mean(dim=-1,keepdim=True)) #mean is apporate to zero
print(out_Norm.var(dim=-1,keepdim=True))  #var is one
```

```python
class LayerNorm(nn.Module):
  def __init__(self, emb_dim):
    super().__init__()
    self.epsilon=1e-5
    self.shift=nn.Parameter(torch.zeros(emb_dim))
    self.scale=nn.Parameter(torch.ones(emb_dim))
  def forward(self,inp):
    mean = inp.mean(dim=-1, keepdim=True) 
    var = inp.var(dim=-1, keepdim=True)
    input_Norm=(inp-mean)/torch.sqrt(var+self.epsilon)
    return input_Norm*self.scale+self.shift
```

```python
LN=LayerNorm(out.shape[1])
out_norm=LN.forward(out)
print(out_norm.mean(dim=-1,keepdim=True))
print(out_norm.var(dim=-1,keepdim=True))
```
