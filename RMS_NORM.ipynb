```python
val=x**2
means=val.mean()
sq=np.sqrt(means)
x_norm=x/sq
print(x_norm.mean())
print(x_norm.std())
```

```python
print(np.sqrt((x_norm**2).mean()))  # will print exactly 1.0
```

```python
import torch
import torch.nn as nn
```

```python
class RMS_NORM(nn.Module):
    def __init__(self, dim, epsilon=1e-8):
        super().__init__()
        self.epsilon = epsilon
        self.gamma = nn.Parameter(torch.ones(dim))   

    def forward(self, x):                          
        rms=torch.sqrt(torch.mean((x**2),dim=-1,keepdim=True)+self.epsilon)
        x_norm=x/rms
        return self.gamma*x_norm
```

```python
x = torch.randn(2, 5)
print(f"x without norm :{x}")
norm = RMS_NORM(dim=5)

out = norm(x)
print(f"x after rms norm :{out}")
print(out.shape)  # should be (2, 5)
```
