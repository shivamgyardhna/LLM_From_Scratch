```python
import torch
import torch.nn as nn
```

```python
#sigmoid Linear unit
class Silu(nn.Module):

  def __init__(self):
    super().__init__()
        
  def forward(self, x):
    return x*(1/(1+torch.exp(-x)))
```

# In pytorch

`silu(x)` <br>
`   ↓` <br>
`__call__()` <br>
` ↓` <br>
`forward(x)`


```python
x=torch.tensor([1,2,3,4])
silu=Silu()
silu(x)
```
