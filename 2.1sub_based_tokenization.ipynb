```python
def get_pairs(words):
  dic={}
  for word in words:
    for ch in word:
      if ch in dic.keys():
        dic[ch]+=1
      else:
        dic[ch]=1  
  return dic    
```

```python
from collections import defaultdict
def get_pairs_col(words):
  d=defaultdict(int)
  for word in words:
    for ch in word:
      d[ch]+=1        
  return d 
# dict(sorted(pairs,key=lambda pair :pair[1],reverse=False))
```

```python
def merge(pairs):
  # pass
  pairs=dict(sorted(pairs.items(),key=lambda item:item[1],reverse=True ))
  #merge
  return pairs
  
  
```

```python
words = ["love", "read", "reader", "lover", "run","shivam"]
pairs = get_pairs(words)
pairs_coll=get_pairs_col(words)
p=merge(pairs_coll)
print(p)
```
